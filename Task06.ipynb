```python
print("Duplicate Rows :", df.duplicated().sum())
df.drop_duplicates(inplace=True)
print("Duplicate Rows :", df.duplicated().sum())
```

```python
df["Engagement_Rate"] = (
    df["Total_Engagement"] /
    df["Reach"]
) * 100
df[["Total_Engagement","Reach","Engagement_Rate"]].head()
```

```python
df["ROI"] = (
    (df["Revenue"] - df["Spend"])
    /
    df["Spend"]
) * 100
df[["Spend","Revenue","ROI"]].head()
```

```python
plt.figure(figsize=(6,4))

plt.text(
    0.05,
    0.5,
    "Engagement Rate = (Total Engagement / Reach) × 100",
    fontsize=14,
    bbox=dict(facecolor="lightblue")
)

plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.text(
    0.05,
    0.5,
    "ROI = ((Revenue - Spend) / Spend) × 100",
    fontsize=14,
    bbox=dict(facecolor="lightgreen")
)

plt.axis("off")

plt.show()
```

```python
total_campaigns = df["Campaign_ID"].nunique()

print("Total Campaigns :", total_campaigns)
```

```python
total_reach = df["Reach"].sum()

print("Total Reach :", total_reach)
```

```python
total_revenue = df["Revenue"].sum()

print("Total Revenue :", total_revenue)
```

```python
total_spend = df["Spend"].sum()

print("Total Spend :", total_spend)
```

```python
avg_engagement_rate = df["Engagement_Rate"].mean()

print("Average Engagement Rate :", round(avg_engagement_rate, 2), "%")
```

```python
avg_roi = df["ROI"].mean()

print("Average ROI :", round(avg_roi, 2), "%")
```

```python
cost_per_engagement = (
    df["Spend"].sum() /
    df["Total_Engagement"].sum()
)

print("Cost Per Engagement :", round(cost_per_engagement, 2))
```

```python
conversion_rate = (
    df["Conversions"].sum() /
    df["Clicks"].sum()
) * 100

print("Conversion Rate :", round(conversion_rate, 2), "%")
```

```python
kpi = pd.DataFrame({
    "KPI": [
        "Total Campaigns",
        "Total Reach",
        "Total Revenue",
        "Total Spend",
        "Average Engagement Rate",
        "Average ROI",
        "Cost Per Engagement",
        "Conversion Rate"
    ],
    "Value": [
        total_campaigns,
        total_reach,
        total_revenue,
        total_spend,
        round(avg_engagement_rate, 2),
        round(avg_roi, 2),
        round(cost_per_engagement, 2),
        round(conversion_rate, 2)
    ]
})

kpi
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
platform_engagement = df.groupby("Platform")["Engagement_Rate"].mean()

plt.figure(figsize=(8,5))
platform_engagement.plot(kind="bar", color="skyblue")

plt.title("Platform vs Engagement")
plt.xlabel("Platform")
plt.ylabel("Average Engagement Rate (%)")

plt.show()
```

```python
top_influencers = df.groupby("Influencer")["Revenue"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top_influencers.plot(kind="bar", color="green")

plt.title("Influencer vs Revenue")
plt.xlabel("Influencer")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Spend"], df["ROI"])

plt.title("Campaign Cost vs ROI")
plt.xlabel("Campaign Spend")
plt.ylabel("ROI (%)")

plt.show()
```

```python
audience_performance = df.groupby("Audience_Group")["ROI"].mean()

plt.figure(figsize=(8,5))
audience_performance.plot(kind="bar", color="orange")

plt.title("Audience Group vs Performance")
plt.xlabel("Audience Group")
plt.ylabel("Average ROI (%)")

plt.show()
```
